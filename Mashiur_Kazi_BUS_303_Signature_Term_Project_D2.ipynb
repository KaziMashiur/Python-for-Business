**Task 2.13:** 

- Remove the special values inside each column.

Special characters were identified in the Total Fat (g), Saturated Fat (g), Trans Fat (g), Cholesterol (mg), Sodium (mg), Fiber (g), and Protein (g) columns.

You should have identified the "<5" special value in the Cholesterol (mg) column and the "\xaO" in multiple columns. The "<5" is isolated to the Cholesterol (mg) column, so that special value can be removed by just removing it from that one column. For the "\xaO" special character, you could remove it one column at a time, but there is a more efficient way to do it. You can remove it from all the columns at once. This is probably the best way to do it, but it is not exactly beginner-level, so I am providing the solution for you. I highly recommend that you run this code through an AI program and prompt it to explain it to you in detail:

```python
for col in dnkn2.select_dtypes(include='object').columns:
    dnkn2[col] = (
        dnkn2[col]
        .str.replace('\xa0', '', regex=False)
        .str.strip()
    )
```

````python
# Task 2.13a: Remove the special values inside the Cholesterol column.

# Remove the <5 value from the Cholesterol column

# ENTER CODE BELOW THIS LINE.
dnkn2['Cholesterol (mg)'] = (dnkn2['Cholesterol (mg)'].replace('<5', '5'))
````

````python
# Task 2.13b: Remove the "\xaO" special character from all columns:

# ENTER THE CODE BELOW THIS LINE:

for col in dnkn2.select_dtypes(include='object').columns:
    dnkn2[col] = dnkn2[col].str.replace('\xa0', '', regex=False).str.strip()
````

**Task 2.15:**

- Remove '' empty strings and replace with NaN values.

Now that you removed the special values from the dnkn2 dataframe, the removal of them created a '' entry in your columns. This is an empty string. These must be treated. Although the special character is gone, the empty string still represents missing information. Therefore, the empty string must be converted to a NaN value so Python can properly identify and manage missing data.

The process is to convert the '' to an NaN. You need to import numpy as np and replace the '' entries in the dnkn2 dataframe with nan values.

dnkn2 = dnkn2.replace('', np.nan)

````python
# Task 2.15: Remove '' empty strings and replace with NaN values.

# ENTER CODE BELOW THIS LINE.
import numpy as np

dnkn2 = dnkn2.replace('', np.nan)
````
